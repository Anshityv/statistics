##  Theoretical Questions

### 1. What is a random variable in probability theory?
A **random variable** is a mathematical function that maps outcomes of a random phenomenon to numerical values. It quantifies uncertain events (e.g., let $X$ be the number of heads obtained when tossing a coin twice).

### 2. What are the types of random variables?
* **Discrete Random Variables:** Variables that take finite or countably infinite distinct values (e.g., number of students, rolling a die).
* **Continuous Random Variables:** Variables that can take any infinite real value within a continuous range or interval (e.g., exact height, weight, time).

### 3. What is the difference between discrete and continuous distributions?
* **Discrete Distribution:** Defined by a **Probability Mass Function (PMF)** where probabilities are assigned to specific individual values ($P(X = x) \ge 0$, and $\sum P(X=x) = 1$).
* **Continuous Distribution:** Defined by a **Probability Density Function (PDF)** where the probability at an exact point is 0 ($P(X = x) = 0$), and probabilities are calculated over intervals as the area under the curve ($\int PDF = 1$).

### 4. What are probability distribution functions (PDF)?
A **Probability Density Function (PDF)** represents the relative likelihood of a continuous random variable taking on a specific value. The area under the PDF curve across an interval $[a, b]$ gives the probability that the variable falls within that interval.

### 5. How do cumulative distribution functions (CDF) differ from probability distribution functions (PDF)?
* **PDF $f(x)$:** Gives the probability *density* at a specific value $x$.
* **CDF $F(x)$:** Gives the *accumulated probability* that the random variable $X$ takes a value less than or equal to $x$ ($F(x) = P(X \le x)$). It ranges monotonically from 0 to 1.

### 6. What is a discrete uniform distribution?
A distribution where a finite number of values are equally likely to occur. For $n$ possible outcomes, each outcome has a probability of $P(X = x) = \frac{1}{n}$ (e.g., rolling a fair 6-sided die).

### 7. What are the key properties of a Bernoulli distribution?
* Models a single trial with only two outcomes: Success ($X = 1$) with probability $p$, and Failure ($X = 0$) with probability $q = 1 - p$.
* **Mean (Expected Value):** $E[X] = p$
* **Variance:** $Var(X) = p(1 - p)$

### 8. What is the binomial distribution, and how is it used in probability?
The **Binomial Distribution** models the number of successes in $n$ independent and identical Bernoulli trials with a constant success probability $p$. Formula: $P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$. It is used to model binary outcomes repeated over multiple trials.

### 9. What is the Poisson distribution and where is it applied?
The **Poisson Distribution** models the count of events occurring within a fixed interval of time or space, assuming events occur at a constant average rate $\lambda$ independently of each other. Applied in call center arrivals, traffic count, and website traffic.

### 10. What is a continuous uniform distribution?
A distribution where continuous values within a specified range $[a, b]$ have equal probability density. PDF: $f(x) = \frac{1}{b - a}$ for $a \le x \le b$.

### 11. What are the characteristics of a normal distribution?
* Symmetric, bell-shaped curve centered around the mean $\mu$.
* Mean, median, and mode are all equal ($\mu = \text{Median} = \text{Mode}$).
* Follows the Empirical Rule (68-95-99.7 rule).

### 12. What is the standard normal distribution, and why is it important?
A special normal distribution with a mean of 0 ($\mu = 0$) and standard deviation of 1 ($\sigma = 1$), denoted as $Z \sim N(0, 1)$. It standardizes different normal distributions onto a universal scale using Z-scores.

### 13. What is the Central Limit Theorem (CLT), and why is it critical in statistics?
The **CLT** states that the sampling distribution of the sample mean approaches a normal distribution as sample size $n$ increases ($n \ge 30$), regardless of the shape of the underlying population distribution. It is critical because it allows inferential statistics (confidence intervals, hypothesis testing) without knowing population distribution.

### 14. How does the Central Limit Theorem relate to the normal distribution?
CLT establishes that even if a population distribution is uniform, exponential, or skewed, the distribution of its sample means $\bar{X}$ will be normally distributed around the population mean $\mu$ with standard error $\frac{\sigma}{\sqrt{n}}$.

### 15. What is the application of Z statistics in hypothesis testing?
Z-statistics test hypotheses about population means when population standard deviation $\sigma$ is known and sample size $n \ge 30$. $Z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$.

### 16. How do you calculate a Z-score, and what does it represent?
* Formula: $Z = \frac{X - \mu}{\sigma}$
* Represents how many standard deviations a raw data point $X$ lies above or below the population mean $\mu$.

### 17. What are point estimates and interval estimates in statistics?
* **Point Estimate:** A single numeric value used to estimate a population parameter (e.g., sample mean $\bar{X}$ for $\mu$).
* **Interval Estimate:** A range of values (confidence interval) within which the population parameter is expected to fall with a specified degree of confidence.

### 18. What is the significance of confidence intervals in statistical analysis?
Confidence intervals quantify uncertainty in sample estimates, offering a margin of error and an estimated range for population parameters rather than relying on single point estimates.

### 19. What is the relationship between a Z-score and a confidence interval?
The critical Z-value ($Z_{\alpha/2}$) determines the width/margin of error of the confidence interval ($	ext{CI} = \bar{X} \pm Z_{\alpha/2} \cdot \frac{\sigma}{\sqrt{n}}$).

### 20. How are Z-scores used to compare different distributions?
Z-scores standardize data onto a common scale (mean=0, std=1), allowing direct comparison of observations across variables measured on different scales (e.g., SAT vs ACT scores).

### 21. What are the assumptions for applying the Central Limit Theorem?
1. Random sampling.
2. Independent observations (sample size $n < 10\%$ of population size).
3. Sufficiently large sample size ($n \ge 30$).

### 22. What is the concept of expected value in a probability distribution?
The long-run weighted average value of a random variable ($E[X] = \sum x P(x)$ or $\int x f(x) dx$).

### 23. How does a probability distribution relate to the expected outcome of a random variable?
The probability distribution provides the probabilities/weights assigned to all possible values, which are used directly to calculate expected outcome ($E[X]$).

##  Practical Questions (Python Code)

### Setup & Libraries

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries imported successfully!')
```

### P1 & P2: Random Variable & Discrete Uniform Distribution (PMF)

```python
# P1: Generate random variable
rv_val = np.random.randint(1, 100)
print(f'P1: Generated Random Variable Value = {rv_val}')

# P2: Discrete Uniform Distribution PMF
outcomes = np.arange(1, 7)
pmf = [1/6] * 6
plt.figure(figsize=(6, 4))
plt.stem(outcomes, pmf)
plt.title('Discrete Uniform PMF (6-sided die)')
plt.xlabel('Outcome')
plt.ylabel('Probability')
plt.show()
```

### P3 & P4: Bernoulli Function & Binomial Simulation (n=10, p=0.5)

```python
# P3: Bernoulli PDF/PMF Function
def bernoulli_pmf(k, p):
    return p if k == 1 else (1 - p)

print(f'P3: Bernoulli PMF for P(X=1, p=0.7) = {bernoulli_pmf(1, 0.7)}')

# P4: Binomial Distribution (n=10, p=0.5)
bin_data = np.random.binomial(n=10, p=0.5, size=1000)
plt.figure(figsize=(6, 4))
plt.hist(bin_data, bins=np.arange(-0.5, 11.5, 1), density=True, rwidth=0.8, color='skyblue')
plt.title('Binomial Distribution Simulation (n=10, p=0.5)')
plt.xlabel('Number of Successes')
plt.ylabel('Density')
plt.show()
```

### P5 & P6: Poisson Visualization & Discrete Uniform CDF

```python
# P5: Poisson Distribution (lambda=4)
poisson_data = np.random.poisson(lam=4, size=1000)
plt.figure(figsize=(6, 4))
sns.histplot(poisson_data, discrete=True, color='teal')
plt.title('Poisson Distribution (lambda=4)')
plt.show()

# P6: Discrete Uniform CDF
cdf = np.cumsum(pmf)
plt.figure(figsize=(6, 4))
plt.step(outcomes, cdf, where='post', color='darkorange')
plt.title('Discrete Uniform CDF')
plt.xlabel('Outcome')
plt.ylabel('CDF')
plt.show()
```

### P7 & P8: Continuous Uniform & Normal Distribution Histogram

```python
# P7: Continuous Uniform
unif_data = np.random.uniform(low=0, high=10, size=1000)
plt.figure(figsize=(6, 4))
sns.kdeplot(unif_data, fill=True, color='purple')
plt.title('Continuous Uniform Distribution [0, 10]')
plt.show()

# P8: Normal Distribution Histogram
norm_data = np.random.normal(loc=50, scale=10, size=1000)
plt.figure(figsize=(6, 4))
sns.histplot(norm_data, kde=True, color='coral')
plt.title('Normal Distribution Histogram')
plt.show()
```

### P9 & P10: Z-score Function & CLT Implementation (Non-Normal Data)

```python
# P9: Z-Score Function & Plot
def calculate_zscores(data):
    return (data - np.mean(data)) / np.std(data)

z_scores = calculate_zscores(norm_data)
plt.figure(figsize=(6, 4))
sns.histplot(z_scores, kde=True, color='green')
plt.title('Calculated Z-Scores')
plt.show()

# P10: CLT Implementation on Exponential (Non-Normal) Distribution
exp_pop = np.random.exponential(scale=2.0, size=10000)
sample_means = [np.mean(np.random.choice(exp_pop, size=40)) for _ in range(1000)]

plt.figure(figsize=(6, 4))
sns.histplot(sample_means, kde=True, color='magenta')
plt.title('CLT Sample Means (from Exponential Population)')
plt.show()
```

### P15 & P16: CLT Verification & Standard Normal Distribution

```python
# P15: CLT Verification from Normal Distribution
norm_sample_means = [np.mean(np.random.normal(0, 1, 30)) for _ in range(1000)]
plt.figure(figsize=(6, 4))
sns.histplot(norm_sample_means, kde=True, color='indigo')
plt.title('P15: CLT Verification with Sample Means')
plt.show()

# P16: Standard Normal PDF Plot
x_vals = np.linspace(-4, 4, 200)
pdf_vals = stats.norm.pdf(x_vals, mean:=0, std:=1)
plt.figure(figsize=(6, 4))
plt.plot(x_vals, pdf_vals, color='red', label='Standard Normal PDF')
plt.title('P16: Standard Normal Distribution (N(0,1))')
plt.legend()
plt.show()
```

### P17 & P18: Binomial Probabilities & Z-Score Comparison

```python
# P17: Binomial Probabilities
n_trials, p_succ = 10, 0.6
k_vals = np.arange(0, n_trials + 1)
probs = stats.binom.pmf(k_vals, n_trials, p_succ)
print(f'P17: Binomial Probabilities for n=10, p=0.6:\n{probs}')

# P18: Z-Score Comparison
val = 75
mean, std = 60, 10
z_val = (val - mean) / std
print(f'P18: Z-score for {val} (mean={mean}, std={std}) = {z_val}')
```

### P19, P20 & P21: Hypothesis Testing & Confidence Intervals

```python
# P19: Z-Test Hypothesis Testing
sample = np.random.normal(loc=102, scale=15, size=50)
z_stat = (np.mean(sample) - 100) / (15 / np.sqrt(50))
p_val = 2 * (1 - stats.norm.cdf(abs(z_stat)))
print(f'P19: Z-Statistic = {z_stat:.4f}, P-Value = {p_val:.4f}')

# P20 & P21: Confidence Interval for Mean (95% CI)
mean_est = np.mean(sample)
sem = stats.sem(sample)
ci = stats.norm.interval(0.95, loc=mean_est, scale=sem)
print(f'P20/P21: 95% Confidence Interval for Mean = {ci}')
```

### P22, P23 & P24: PDF/CDF Functions & Expected Value

```python
# P22: Normal PDF Visualization
x = np.linspace(20, 80, 200)
plt.figure(figsize=(6, 4))
plt.plot(x, stats.norm.pdf(x, 50, 10), color='brown')
plt.title('P22: Normal Distribution PDF')
plt.show()

# P23: Poisson CDF
poisson_cdf = stats.poisson.cdf(np.arange(0, 10), mu=4)
print(f'P23: Poisson CDF (lambda=4, k=0..9):\n{poisson_cdf}')

# P24: Continuous Uniform Expected Value
unif_sample = np.random.uniform(a:=5, b:=15, 10000)
exp_val_calc = (a + b) / 2
print(f'P24: Calculated Expected Value = {exp_val_calc}, Sample Mean = {np.mean(unif_sample):.4f}')
```

### P25, P26, P27 & P28: Data Comparison & Summary Statistics

```python
# P25: Standard Deviation Comparison
d1, d2 = np.random.normal(0, 5, 500), np.random.normal(0, 15, 500)
plt.figure(figsize=(6, 4))
sns.kdeplot(d1, label=f'Dataset 1 (std={np.std(d1):.2f})')
sns.kdeplot(d2, label=f'Dataset 2 (std={np.std(d2):.2f})')
plt.title('P25: Standard Deviation Comparison')
plt.legend()
plt.show()

# P26: Range and IQR
norm_dist = np.random.normal(100, 15, 1000)
data_range = np.ptp(norm_dist)
iqr_val = stats.iqr(norm_dist)
print(f'P26: Range = {data_range:.2f}, IQR = {iqr_val:.2f}')

# P27: Z-score Normalization
norm_transformed = stats.zscore(norm_dist)
print(f'P27: Transformed Mean = {np.mean(norm_transformed):.4f}, Std = {np.std(norm_transformed):.4f}')

# P28: Skewness & Kurtosis
skew_val = stats.skew(norm_dist)
kurt_val = stats.kurtosis(norm_dist)
print(f'P28: Skewness = {skew_val:.4f}, Kurtosis = {kurt_val:.4f}')
```
