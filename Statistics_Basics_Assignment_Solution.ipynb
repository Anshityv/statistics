### 1. Explain the different types of data (qualitative and quantitative) and provide examples of each. Discuss nominal, ordinal, interval, and ratio scales.
* **Qualitative (Categorical) Data:** Data that describes characteristics, categories, or labels and cannot be measured numerically.
  * *Example:* Eye color (Blue, Brown, Green), Marital status (Single, Married).
* **Quantitative (Numerical) Data:** Data that represents numbers, counts, or measurable quantities.
  * *Example:* Height (175 cm), Temperature (25°C), Number of items sold (150).

**Levels of Measurement / Scales:**
* **Nominal Scale:** Categories with no quantitative or natural ordering (e.g., Blood types: A, B, AB, O).
* **Ordinal Scale:** Categories with a meaningful order or ranking, but exact differences between values cannot be measured (e.g., Customer satisfaction ratings: Poor, Average, Excellent).
* **Interval Scale:** Numeric scales where the order and exact differences between values are meaningful, but there is *no true zero* point (e.g., Temperature in Celsius or Fahrenheit; 0°C does not mean 'no temperature').
* **Ratio Scale:** Numeric scales possessing all properties of interval scale plus a *true absolute zero* point, meaning zero represents complete absence of the variable (e.g., Weight, Height, Income, Age).

### 2. What are the measures of central tendency, and when should you use each? Discuss the mean, median, and mode with examples and situations where each is appropriate.
* **Mean (Average):** Calculated by summing all values and dividing by the total count. 
  * *When to use:* Symmetric distributions without extreme outliers (e.g., tracking average monthly test scores in a stable class).
* **Median (Middle Value):** The middle value when data is sorted in ascending/descending order.
  * *When to use:* Skewed distributions or datasets with extreme outliers, because outliers do not distort the median (e.g., median household income in a city).
* **Mode (Most Frequent Value):** The value that appears most often in the dataset.
  * *When to use:* Categorical data to find the most popular category or item (e.g., most frequently purchased shoe size or favorite mobile brand).

### 3. Explain the concept of dispersion. How do variance and standard deviation measure the spread of data?
* **Dispersion** measures how spread out or scattered a set of data points is around their central tendency (like the mean).
* **Variance ($s^2$):** The average of the squared differences from the mean. Squaring makes all differences positive and heavily penalizes larger deviations.
* **Standard Deviation ($s$):** The square root of the variance. It returns the measure to the original units of the data, making it intuitive to understand how far data points typically deviate from the mean.

### 4. What is a box plot, and what can it tell you about the distribution of data?
* A **box plot** (or box-and-whisker plot) is a graphical representation of data summary based on a five-number summary: Minimum, First Quartile ($Q_1$), Median ($Q_2$), Third Quartile ($Q_3$), and Maximum.
* **What it tells you:**
  * **Center and Spread:** The median line shows the center; the box size (IQR) shows the middle 50% spread.
  * **Skewness:** If the median line is closer to one quartile or whiskers are unequal, the data is skewed.
  * **Outliers:** Individual points plotted beyond the whiskers represent potential anomalies or outliers.

### 5. Discuss the role of random sampling in making inferences about populations.
* **Random sampling** ensures that every member of a target population has an equal and independent chance of being selected in the sample.
* **Role in Inferences:** It eliminates selection bias and ensures that the sample is representative of the entire population. This allows researchers to use sample statistics (like sample mean or proportion) to reliably estimate population parameters and perform hypothesis testing with measurable confidence levels.

### 6. Explain the concept of skewness and its types. How does skewness affect the interpretation of data?
* **Skewness** measures the asymmetry of the probability distribution of a real-valued random variable about its mean.
* **Types:**
  * **Positive (Right) Skew:** The tail is longer on the right side; the mean is greater than the median (e.g., personal incomes).
  * **Negative (Left) Skew:** The tail is longer on the left side; the mean is less than the median (e.g., retirement age).
  * **Zero Skew:** Perfectly symmetrical distribution (Mean = Median = Mode).
* **Impact:** Skewness causes the mean to be pulled toward the tail, making the median a safer measure of central tendency for skewed data.

### 7. What is the interquartile range (IQR), and how is it used to detect outliers?
* **IQR** is the difference between the 75th percentile ($Q_3$) and the 25th percentile ($Q_1$), representing the middle 50% of the data ($IQR = Q_3 - Q_1$).
* **Outlier Detection:** Data points that fall below $Q_1 - 1.5 \times \text{IQR}$ or above $Q_3 + 1.5 \times \text{IQR}$ are typically flagged as outliers.

### 8. Discuss the conditions under which the binomial distribution is used.
The binomial distribution models the number of successes in $n$ independent trials of a binary experiment. Conditions include:
1. **Fixed number of trials ($n$):** The experiment is repeated a fixed number of times.
2. **Binary outcomes:** Each trial has only two possible outcomes (Success/Failure, Yes/No).
3. **Constant probability ($p$):** The probability of success remains the same for each trial.
4. **Independence:** The outcome of any trial does not affect the outcome of subsequent trials.

### 9. Explain the properties of the normal distribution and the empirical rule (68-95-99.7 rule).
* **Properties:** Symmetric, bell-shaped curve; mean, median, and mode are all equal and located at the center; asymptotic tails.
* **Empirical Rule:**
  * **68%** of data falls within $\pm 1$ standard deviation of the mean ($\mu \pm \sigma$).
  * **95%** falls within $\pm 2$ standard deviations ($\mu \pm 2\sigma$).
  * **99.7%** falls within $\pm 3$ standard deviations ($\mu \pm 3\sigma$).

### 11. Explain what a random variable is and differentiate between discrete and continuous random variables.
* **Random Variable:** A mathematical formalization of a quantity or object which depends on random events.
* **Discrete Random Variable:** Takes countable distinct values (e.g., number of students in a classroom: 0, 1, 2, ...).
* **Continuous Random Variable:** Takes any numerical value within a continuous range or interval (e.g., exact height or weight).

##  Practical Programming & Calculation Section

### Setup & Libraries

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries imported successfully!')
```

### Practical Code Solutions for Statistical Calculations

```python
# Example Dataset for Demonstration (e.g., Study Hours vs Exam Scores)
data = {
    'Study_Hours': [2.5, 3.0, 3.5, 4.0, 4.5, 5.0, 5.5, 6.0, 6.5, 7.0],
    'Exam_Score': [65, 70, 75, 78, 82, 85, 88, 92, 95, 98]
}
df = pd.DataFrame(data)

# 1. Measures of Central Tendency & Dispersion
mean_score = df['Exam_Score'].mean()
median_score = df['Exam_Score'].median()
mode_score = df['Exam_Score'].mode()[0]
variance_score = df['Exam_Score'].var(ddof=1)
std_score = df['Exam_Score'].std(ddof=1)

print(f"--- Central Tendency & Dispersion ---")
print(f"Mean Exam Score: {mean_score}")
print(f"Median Exam Score: {median_score}")
print(f"Mode Exam Score: {mode_score}")
print(f"Variance: {variance_score:.2f}")
print(f"Standard Deviation: {std_score:.2f}\n")

# 2. IQR & Outlier Detection Calculation
Q1 = df['Exam_Score'].quantile(0.25)
Q3 = df['Exam_Score'].quantile(0.75)
iqr = Q3 - Q1
lower_bound = Q1 - 1.5 * iqr
upper_bound = Q3 + 1.5 * iqr

print(f"--- IQR & Outlier Bounds ---")
print(f"Q1: {Q1}, Q3: {Q3}, IQR: {iqr}")
print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}\n")

# 3. Poisson Probability Calculation (Q10)
lam = 4  # Average 4 calls per hour
k = 2    # Exactly 2 calls
poisson_prob = stats.poisson.pmf(k, lam)
print(f"--- Poisson Probability (Q10) ---")
print(f"Probability of exactly {k} calls when lambda={lam} is: {poisson_prob:.4f}\n")

# 4. Covariance and Correlation Calculation (Q12)
covariance_matrix = np.cov(df['Study_Hours'], df['Exam_Score'])
covariance = covariance_matrix[0, 1]
correlation, p_value = stats.pearsonr(df['Study_Hours'], df['Exam_Score'])

print(f"--- Covariance and Correlation (Q12) ---")
print(f"Covariance: {covariance:.2f}")
print(f"Pearson Correlation Coefficient: {correlation:.4f}")
print(f"Interpretation: Strong positive linear relationship between study hours and exam scores.")
```
