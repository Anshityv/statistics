### Question 1: What is hypothesis testing in statistics?
**Answer:**  
Hypothesis testing is a statistical methodology used to make inferences or decisions about population parameters based on sample data. It involves making an initial claim (the null hypothesis) about a population parameter, collecting sample data, and using statistical evidence to determine whether there is sufficient support to reject the null hypothesis in favor of an alternative claim. It provides a formal framework for testing assumptions, scientific theories, and operational claims while controlling for random sampling error.

---

### Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?
**Answer:**  
* **Null Hypothesis ($H_0$):** Represents the default assumption or baseline condition of no effect, no difference, or no change. It is presumed true until sufficient statistical evidence indicates otherwise (e.g., $H_0: \mu = 50$).
* **Alternative Hypothesis ($H_1$ or $H_a$):** Represents the researcher's claim or hypothesis that there is a statistically significant effect, difference, or change (e.g., $H_1: \mu \neq 50$, $H_1: \mu > 50$, or $H_1: \mu < 50$).
* **Key Difference:** Hypothesis testing focuses strictly on evaluating whether sample evidence is strong enough to reject $H_0$. If evidence is weak, we fail to reject $H_0$; we never "prove" $H_0$ absolute truth.

---

### Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.
**Answer:**  
The **significance level** (denoted by $\alpha$) is the maximum threshold probability of committing a Type I error (rejecting the null hypothesis when it is actually true). Standard chosen values are $\alpha = 0.05$ (5%), $\alpha = 0.01$ (1%), or $\alpha = 0.10$ (10%).

**Role in Decision Making:**
* **Using p-value:**
  * If **$p$-value $\le \alpha$**: Reject the null hypothesis ($H_0$). The sample result is statistically significant.
  * If **$p$-value $> \alpha$**: Fail to reject the null hypothesis ($H_0$). There is insufficient evidence to support $H_1$.
* **Using Critical Values:** If the calculated test statistic (e.g., Z-score or T-score) falls inside the critical rejection region determined by $\alpha$, $H_0$ is rejected.

---

### Question 4: What are Type I and Type II errors? Give examples of each.
**Answer:**  
* **Type I Error (False Positive, $\alpha$):** Occurs when the null hypothesis ($H_0$) is true, but we incorrectly reject it.
  * *Example:* A drug trial concludes a new medication is effective when in reality it has no effect.
  * *Legal Example:* Convicting an innocent person in a criminal trial.
* **Type II Error (False Negative, $\beta$):** Occurs when the null hypothesis ($H_0$) is false, but we fail to reject it.
  * *Example:* A medical test fails to detect a disease in a patient who actually has the disease.
  * *Legal Example:* Acquitting a guilty person due to lack of conclusive evidence.

---

### Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.
**Answer:**  
* **Z-Test:**
  * Used when population standard deviation ($\sigma$) is known OR sample size is large ($n \ge 30$).
  * Relies on the standard normal distribution ($Z \sim N(0,1)$).
* **T-Test:**
  * Used when population standard deviation ($\sigma$) is unknown AND sample size is small ($n < 30$).
  * Relies on Student's t-distribution, which has heavier tails to account for the additional uncertainty of estimating $\sigma$ with sample standard deviation ($s$).

### Question 6: Write a Python program to generate a binomial distribution with $n=10$ and $p=0.5,$ then plot its histogram.

```python
# Set random seed for reproducibility
np.random.seed(42)

# Simulation parameters
n = 10
p = 0.5
num_simulations = 1000

# Generate binomial distribution using random choices (representing coin flips)
binomial_data = [np.sum(np.random.choice([0, 1], size=n, p=[1-p, p])) for _ in range(num_simulations)]

# Plot histogram
plt.figure(figsize=(8, 5))
plt.hist(binomial_data, bins=np.arange(-0.5, n + 1.5, 1), density=True, rwidth=0.8, color='skyblue', edgecolor='black')
plt.title('Binomial Distribution Simulation (n=10, p=0.5)')
plt.xlabel('Number of Successes')
plt.ylabel('Probability Density')
plt.xticks(range(0, n + 1))
plt.grid(axis='y', alpha=0.75)
plt.show()
```

### Question 7: Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.

```python
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6, 
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5, 
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9, 
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

# Hypotheses:
# H0: mu = 50.0 (Population mean is 50.0)
# H1: mu != 50.0 (Two-tailed test)
mu_0 = 50.0
alpha = 0.05

sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)  # Sample standard deviation
n = len(sample_data)
standard_error = sample_std / np.sqrt(n)

# Calculate Z-statistic
z_stat = (sample_mean - mu_0) / standard_error

# Calculate two-tailed p-value
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print(f"Sample Mean: {sample_mean:.4f}")
print(f"Sample Size (n): {n}")
print(f"Z-statistic: {z_stat:.4f}")
print(f"p-value: {p_value:.4f}")

if p_value <= alpha:
    print(f"Conclusion: Reject Null Hypothesis (H0). There is significant evidence that mu != {mu_0}.")
else:
    print(f"Conclusion: Fail to Reject Null Hypothesis (H0). Insufficient evidence to suggest population mean differs from {mu_0}.")
```

### Question 8: Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

```python
# Simulate normal data
np.random.seed(42)
simulated_data = np.random.normal(loc=100, scale=15, size=100)

# Calculate 95% Confidence Interval for mean
mean_val = np.mean(simulated_data)
sem = stats.sem(simulated_data)
ci_lower, ci_upper = stats.norm.interval(0.95, loc=mean_val, scale=sem)

print(f"Sample Mean: {mean_val:.2f}")
print(f"95% Confidence Interval for Mean: ({ci_lower:.2f}, {ci_upper:.2f})")

# Plot using Matplotlib
plt.figure(figsize=(8, 5))
plt.hist(simulated_data, bins=15, density=True, alpha=0.6, color='mediumpurple', edgecolor='black', label='Simulated Data')
plt.axvline(mean_val, color='red', linestyle='dashed', linewidth=2, label=f'Mean: {mean_val:.2f}')
plt.axvline(ci_lower, color='green', linestyle='dotted', linewidth=2, label=f'95% CI Lower: {ci_lower:.2f}')
plt.axvline(ci_upper, color='green', linestyle='dotted', linewidth=2, label=f'95% CI Upper: {ci_upper:.2f}')
plt.title('Normal Distribution Simulation with 95% Confidence Interval')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Question 9: Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.

```python
def calculate_and_plot_zscores(data):
    """
    Calculates Z-scores for a dataset and plots standard histogram.
    """
    data = np.array(data)
    mean = np.mean(data)
    std = np.std(data, ddof=0)
    z_scores = (data - mean) / std
    
    # Plot standard histogram
    plt.figure(figsize=(8, 5))
    sns.histplot(z_scores, kde=True, color='seagreen')
    plt.axvline(0, color='red', linestyle='--', label='Mean (Z = 0)')
    plt.title('Histogram of Standardized Z-Scores')
    plt.xlabel('Z-Score (Standard Deviations from Mean)')
    plt.ylabel('Frequency')
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()
    
    return z_scores

# Run function on simulated data
z_results = calculate_and_plot_zscores(simulated_data)

print("""
--- Z-Score Interpretation ---
A Z-score measures exactly how many standard deviations a raw data point lies away from the mean of the dataset.
- A Z-score of 0 indicates the value is equal to the mean.
- A positive Z-score (e.g., +1.5) means the value is 1.5 standard deviations ABOVE the mean.
- A negative Z-score (e.g., -2.0) means the value is 2.0 standard deviations BELOW the mean.
""")
```
